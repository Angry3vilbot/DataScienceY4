# Data Summary

This notebook provides a summary of the Steam Games dataset.

## 1. Import Libraries and Load the Dataset

```python
import numpy as np
import pandas as pd

dataset = pd.read_csv('steam_games.csv')
```

## 2. Dataset Overview
### 2.1 Shape of the Dataset

```python
print(f"Dataset Shape: {dataset.shape}")
print(f"Number of Columns: {dataset.shape[1]}")
print(f"Number of Rows: {dataset.shape[0]}")
```

### 2.2 Column Names

```python
print("Column Names")
for col in dataset.columns:
    print(f"- {col}")
```

### 2.3 Data Types

```python
print("Data Types")
dataset.dtypes
```

### 2.4 Missing Values

```python
print("Missing Values")
missing_values = dataset.isnull().sum()
missing_values[missing_values > 0]
```

### 2.5 Example Rows

```python
print("Example Rows")
dataset.head()
```

## 3. Summary Statistics
### 3.1 Numerical Columns

```python
print("Summary Statistics for Numerical Columns")
numerical_summary = dataset.describe()
numerical_summary
```

### 3.2 Categorical Columns

```python
print("Summary Statistics for Categorical Columns")
categorical_summary = dataset.describe(include=['object'])
categorical_summary
```
