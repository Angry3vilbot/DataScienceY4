### Loading the `.json` format of Steam Games Dataset and converting to the `.csv` format

```python
import numpy as np
import pandas as pd

json_path = "./games.json"
```

```python
def convert_dict_to_string(dict_object):
    if not dict_object:
        return np.nan
    return ",".join(dict_object.keys())


def convert_list_to_string(entries):
    if not entries:
        return np.nan
    return ",".join(entries)


def read_convert_json_dataset():
    _df = pd.read_json(json_path)
    _df = _df.T
    _df['AppID'] = _df.index

    for col in _df.columns.values:
        sample = _df.loc[546560, col]

        if isinstance(sample, dict):
            _df[col] = _df[col].apply(
                lambda x: convert_dict_to_string(x)
                if isinstance(x, dict) else x
            )

        elif isinstance(sample, list):
            if not sample or isinstance(sample[0], str):
                _df[col] = _df[col].apply(
                    lambda x: convert_list_to_string(x)
                    if isinstance(x, list) else x
                )

    _df = _df.reset_index(drop=True).set_index("AppID")
    return _df


df = read_convert_json_dataset()
```

Remove empty lists from the `tags` and `packages` columns and replace them with NaN values.

```python
df["tags"] = df["tags"].apply(
    lambda x: np.nan if isinstance(x, list) and len(x) == 0 else x
)
df["packages"] = df["packages"].apply(
    lambda x: np.nan if isinstance(x, list) and len(x) == 0 else x
)
```
